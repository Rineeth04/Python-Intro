```python
#Interface in python 
from abc import ABC,abstractmethod

class payment(ABC):

    @abstractmethod
    def pay(self):
        pass

class UPI(payment):
    def pay(self):
        print("UPI payment successfull")

class creditcard(payment):
    def pay(self):
        print("credit card payment is successfull")


# use that payment options
payments = [UPI(),creditcard()]

for payment in payments:
    payment.pay()
```
