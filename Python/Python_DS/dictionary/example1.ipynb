```python
# dictionary in python are the key-value pairs
# creating a dictionary
dict ={"name":"Rineeth","Age":20,"City":"Hyderabad"}
print(dict)
```

```python
# Adding and updating key-value pairs in a dictionary
dict["Country"] = "India" # Adding a new key-value pair
dict["Age"] = 21 # Updating the value of an existing key
print(dict)
```

```python
#iterate through item in the dictionary
d ={"a":1,"b":2,"c":3}
for key in d:
    print(key) # prints the keys of the dictionary
```

```python
data ={"name":"Rineeth","Age":20,"City":"Hyderabad"}
for key, value in data.items():
    print(key, ":", value) # prints the key-value pairs of the dictionary
```
