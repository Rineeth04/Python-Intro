```python
#List in python are mutable set of values where which can be changed and allows duplicate values. List is ordered collection of items which are enclosed in square brackets [] and separated by commas. List can contain any type of data such as integers, strings, floats, etc. List can also contain other lists as elements. List is a built-in data structure in python which provides various methods to manipulate the list such as append(), insert(), remove(), pop(), etc.
l1=["Rineeth",22,"Male,21.0"]
print(l1)
```

```python
# Tuple methods
# append()
l1 =[9,7,6,2]
l1.append(7)
print(l1)
```

```python
# insert() in list
l1.insert(1,"Python")
print(l1)
print(type(l1))
```

```python
# remove() in list
list = [1,2,3,4,5]
list.remove(4)
print("After removing the list 4:",list)
```

```python
# pop() method in list
list.pop(2)
print("After popping the list 2:",list)
```

```python
# sort() method in list
list =[9,7,10,13,1]
list.sort()
print("After sorting the list:",list)
```

```python
#count() method in list
list.count(7)
```

```python
# LIst comphrehension
squares = [x**2 for x in range(10)]
print(squares)
```

```python
# largest number in the list
list =[10,5,50,77,90]
res = max(list)
print("The largest number in the list is:",res)
```

```python
# Remove duplicate numbers in the list
num = [1,2,3,4,5,2,3,6,7]
for i in num: 
        # Remove duplicate numbers in the list
        num = [1,2,3,4,5,2,3,6,7]
        for i in num[:]:
            if num.count(i) > 1:
                num.remove(i)
        print("After removing the duplicate numbers in the list:", num)
#print("After removing the duplicate numbers in the list:",list)
```
