```python
# String in python are immutable, which means that once a string is created, it cannot be modified. However, you can create a new string by concatenating or slicing existing strings. Here are some examples of how to work with strings in Python:
```

```python
name ="Rineeth"
Gender ="Male"
print(name)
print(Gender)

# Accessing characters in a string
print(name[0]) # Output: R
print(name[1]) # Output: i
print(name[-1]) # Output: h
print(name[0:3]) # Output: Rin
```

```python
# slicing of the strings
str1 ="Hello! python"
print(str1[0:6])
print(str1[7:13:2])
```

```python
# concenating strings
str1 = "Welcome to Python world"
str2 = "Let's learn Python together"
str3 = str1 + " " + str2
print(str3)
```

```python
# python Built-in String methods
name = "Welcome to Python world "
print(len(name))
print("capitalized:", name.capitalize())
print("lowercase:", name.lower())
print("uppercase:", name.upper())
print("title:", name.title())
print("swapcase:", name.swapcase())
print("count of 'o':", name.count('o'))
print("strip:", name.strip())
```

```python
# Find and replace string methods
name = "Python is great programming language"
print("original string:", name)

update =name.replace("Python" ,"Java")
print("update name is:,", update)

print(name.count("g"))

print(name.index("Python"))
print(name.find("n"))

print(name.startswith("Python"))
print(name.endswith("python"))
```

```python
# repetetion of a string
s1 = "Hello Python"
print(s1 * 4)
```

```python
# Formatting the string
# % operator  
# format() method
# f-string
# string template

name = "Rineeth"
age = 22
print("My name is {} and I am {} years old.".format(name, age))

print(f"My name is {name} and I am {age} years old.")

print("My name is %s and I am %d years old." % (name, age))
```

```python
from string import Template
template = Template("My name is $name and I am $age years old.")
print(template.substitute(name=name, age=age))
```

```python
# split and join methods in strings
name = " Python Language "
print("original string is:", name)

# removes all leading and trailing whitespace characters
print("stripped string is:", name.strip())
print("left stripped string is:", name.lstrip())
print("right stripped string is:", name.rstrip())

# split the string into a list of words
print(name.split())
print(name.rsplit)
print(name.splitlines())
print(name.split("a"))
```

```python
# Boolean methods in strings
name = "Rineeth"
age = 22
print(name.isalpha()) # True
print(name.isdigit()) # False
print(name.islower()) # False
print(name.isupper()) # False
print(name.istitle()) # False
print(name.isalnum()) # True
```
