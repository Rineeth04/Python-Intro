```python
# common elements in two lists
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
common_elements = set(list1) & set(list2)
print("Common elements in the two lists are:", common_elements)
```

```python
common=[]
for i in list1:
    if i in list2:
        common.append(i)
print("Common elements in the two lists are:", common)
```

```python
# FInd duplicate elements in the list
list1 =[1,2,1,4,5,3,5,6,0]
dup=[]
for i in list1:
    if list1.count(i)>1 and i not in dup:
        dup.append(i)
print("Duplicate elements in the list are:", dup)
```

```python
# 1st largeset number in the list
list =[80,90,66,81,77,99]
large=list[2]
for i in list:
    if i>large:
        large=i
print("The largest number in the list is:",large)
```

```python
# 2md largest number in the list
list =[80,90,66,81,77,99]
sec_large = list[0]
for i in list:
    if i > sec_large and i < large:
        sec_large = i
print("The second largest number in the list is:", sec_large)
```
