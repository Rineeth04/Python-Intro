```python
# Find the no of vowels in a string
str = "Hello WELcome to the world of Python programming"
vowels = "aeiou"
count = 0
res = str.lower()
for i in str:
    if i in vowels:
        count += 1
print("The number of vowels in the string is:", count)
```

```python
# string comparision
str1 = "Hello"
str2 = "Hello1"
if str1 == str2:
    print("Both strings are equal")
else:
    print("Both strings are not equal")
```

```python
# check if a string is a palindrome
str1="madama"
if str1 == str1[::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```

```python
# count frequency of each character
s1 = "Hello World"
freq ={}
for i in s1:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
print(freq)
```

```python
# Non-repetitive characters in a string
s1 = "Hello World"
non_rep = []
for i in s1:
    if s1.count(i) == 1:
        non_rep.append(i)
print("Non-repetitive characters in the string are:", non_rep)
```

```python
# remove duplicate characters in a string
s1 = "python is good language , python is easy to learn"
res = " "
for i in s1:
    if i not in res:
        res += i
print("String after removing duplicate characters:", res)
```

```python
# find duplicate characters in a string
str1 ="language"
dup = set()
for i in str1:
    if str.count(i)>1:
        dup.add(i)
print("Duplicate characters in the string are:", dup)
```

```python
text = "PyThOn"

upper = 0
lower = 0

for ch in text:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1

print("Upper:", upper)
print("Lower:", lower)
```

```python
# reverse a string
s1 = "python language"
rev = s1[::-1]
print("Reversed string is:", rev)

words = s1.split()
print("".join(words[::-1]))
```
